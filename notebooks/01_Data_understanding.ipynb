# 01. DATA UNDERSTANDING & STRUCTURAL INTEGRITY CHECK
**Hệ thống Phân tích Dữ liệu Khách hàng và Dự đoán Quyết định Gửi tiền Tiết kiệm tại Ngân hàng**
---
Nội dung notebook này thực hiện:
1. Nạp tập dữ liệu gốc `Bank_Marketing_Dataset.csv` (UCI Machine Learning Repository).
2. Chuẩn hóa tên cột sang thuật ngữ nghiệp vụ ngân hàng.
3. Kiểm tra tính toàn vẹn: Missing values (NaN), Duplicate records.
4. Đánh giá phân phối và mức độ mất cân bằng lớp của biến mục tiêu `deposit` (Class).


```python
import pandas as pd
import numpy as np

# 1. Đọc dữ liệu
df = pd.read_csv('../dataset/Bank_Marketing_Dataset.csv')
print(f'Kích thước dữ liệu gốc: {df.shape[0]:,} dòng, {df.shape[1]} cột')
df.head()
```

```python
# 2. Chuẩn hóa tên cột sang thuật ngữ ngân hàng
column_mapping = {
    'V1': 'age', 'V2': 'job', 'V3': 'marital', 'V4': 'education',
    'V5': 'default', 'V6': 'balance', 'V7': 'housing', 'V8': 'loan',
    'V9': 'contact', 'V10': 'day', 'V11': 'month', 'V12': 'duration',
    'V13': 'campaign', 'V14': 'pdays', 'V15': 'previous', 'V16': 'poutcome',
    'Class': 'deposit'
}
df = df.rename(columns=column_mapping)
df['deposit'] = df['deposit'].map({1: 'no', 2: 'yes'})
df['target'] = (df['deposit'] == 'yes').astype(int)
df.info()
```

```python
# 3. Kiểm tra giá trị khuyết và trùng lặp
print('Số lượng ô khuyết thiếu (NaN):', df.isnull().sum().sum())
print('Số dòng trùng lặp (Duplicates):', df.duplicated().sum())

# 4. Đánh giá mức độ mất cân bằng lớp mục tiêu
dist = df['target'].value_counts()
dist_pct = df['target'].value_counts(normalize=True) * 100
print('\nPhân bố nhãn mục tiêu:')
print(pd.DataFrame({'Số lượng': dist, 'Tỷ lệ (%)': dist_pct.round(2)}))
print(f'\nTỷ lệ mất cân bằng (No / Yes): {dist[0] / dist[1]:.2f} : 1')
```
