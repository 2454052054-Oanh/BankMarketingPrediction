**Giải thích mô hình bằng SHAP (SHapley Additive exPlanations) và đề xuất kinh doanh**
---
Nội dung notebook này thực hiện:
1. Khởi tạo `shap.TreeExplainer` trên mô hình Gradient Boosting tối ưu (XGBoost).
2. Trích xuất SHAP Beeswarm Summary Plot (chiều hướng tác động của từng đặc trưng).
3. Trích xuất Global Feature Importance (|mean SHAP value|).
4. Phân tích SHAP cục bộ (Local Waterfall interpretation) cho từng quyết định cá thể.
5. Tổng kết các khuyến nghị nghiệp vụ dành cho nhà quản trị ngân hàng.


```python
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import shap

best_model = joblib.load('../dataset/model/best_model.pkl')
X_test_scaled = joblib.load('../dataset/processed/X_test_scaled.pkl')

# Tính toán SHAP values
explainer = shap.TreeExplainer(best_model)
sample_X = X_test_scaled.iloc[:500]
shap_values = explainer.shap_values(sample_X)

plt.figure(figsize=(10, 8))
shap.summary_plot(shap_values, sample_X, show=True)
```
