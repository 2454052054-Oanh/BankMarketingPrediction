# 03. EXPLORATORY DATA ANALYSIS (EDA) & UNSUPERVISED CLUSTERING
**Phân tích khám phá dữ liệu, trực quan hóa và phân cụm khách hàng K-Means**
---
Nội dung notebook này thực hiện:
1. Trực quan hóa ma trận tương quan Pearson.
2. Phân tích ảnh hưởng của nhân khẩu học và các khoản nợ vay đến quyết định gửi tiền.
3. Phân cụm khách hàng K-Means với phương pháp xác định K tối ưu (Elbow Method).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

df_clean = pd.read_csv('../dataset/cleaned/bank_cleaned.csv')
print('Tập dữ liệu đã làm sạch:', df_clean.shape)
```

```python
# Phân cụm khách hàng K-Means theo tuổi và số dư tài khoản
scaler_km = StandardScaler()
X_cluster = scaler_km.fit_transform(df_clean[['age', 'balance']])

inertias = []
for k in range(1, 8):
    km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(X_cluster)
    inertias.append(km.inertia_)

# K-Means với K=3 tối ưu
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_clean['cluster'] = kmeans.fit_predict(X_cluster)
print('Tọa độ tâm cụm (Centroids quy về thang đo gốc):')
print(scaler_km.inverse_transform(kmeans.cluster_centers_))
```
